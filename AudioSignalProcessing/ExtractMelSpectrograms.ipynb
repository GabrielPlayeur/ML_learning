```python
import librosa
import librosa.display
import IPython.display as ipd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#load audio files
BASE_PATH = './audio'

sca_path = f"{BASE_PATH}/scale.wav"

scale, sr = librosa.load(sca_path)
```

```python
FRAME_SIZE=2048
HOP_SIZE=512
```

```python
# TEST LIBROSA MEL
filter_banks = librosa.filters.mel(n_fft=FRAME_SIZE, sr=22050, n_mels=10)
filter_banks.shape #(nb bands, FRAME_SIZE/2 +1)
```

```python
plt.figure(figsize=(25,10))
librosa.display.specshow(filter_banks, sr=sr, x_axis='linear')
plt.colorbar(format="%+2.f")
plt.show()
# Color equal to weight
```

```python
# EXTRACT MEL SPEC
MEL_BANDS=40

mel_spec = librosa.feature.melspectrogram(y=scale, sr=sr, n_fft=FRAME_SIZE, hop_length=HOP_SIZE, n_mels=MEL_BANDS)
mel_spec.shape
log_mel_spec = librosa.power_to_db(mel_spec)
```

```python
plt.figure(figsize=(25,10))
librosa.display.specshow(log_mel_spec, sr=sr, x_axis='time', y_axis="mel")
plt.colorbar(format="%+2.f")
plt.show()
```
