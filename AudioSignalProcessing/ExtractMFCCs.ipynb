```python
import librosa
import librosa.display
import IPython.display as ipd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#load audio files
BASE_PATH = './audio'

deb_path = f"{BASE_PATH}/debussy.wav"
deb, sr = librosa.load(deb_path)
deb.shape
```

```python
N_MFCC=13
```

```python
mfccs = librosa.feature.mfcc(y=deb, n_mfcc=N_MFCC, sr=sr)
mfccs.shape
```

```python
# VISUAL MFCCs
def plot(sig, sr):
    plt.figure(figsize=(25,10))
    librosa.display.specshow(sig, 
                            sr=sr, 
                            x_axis='time', 
                            )
    plt.colorbar(format="%+2.f")
    plt.show()
```

```python
plot(mfccs, sr=sr)
```

```python
# Calculate Delta/Delta2 MFCCs

d_mfccs = librosa.feature.delta(mfccs)
d_mfccs.shape
d2_mfccs = librosa.feature.delta(mfccs, order=2)
d2_mfccs.shape
```

```python
plot(d_mfccs, sr)
```

```python
plot(d2_mfccs, sr)
```

```python
comprehensive_mfccs = np.concatenate((mfccs, d_mfccs, d2_mfccs))
comprehensive_mfccs.shape
```

```python
plot(comprehensive_mfccs, sr)
```
