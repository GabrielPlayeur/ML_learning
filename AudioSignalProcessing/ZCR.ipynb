```python
import librosa
import librosa.display
import IPython.display as ipd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#load audio files
BASE_PATH = './audio'

deb_path = f"{BASE_PATH}/debussy.wav"
duke_path = f"{BASE_PATH}/duke.wav"
redhot_path = f"{BASE_PATH}/redhot.wav"

deb, _ = librosa.load(deb_path)
duk, _ = librosa.load(duke_path)
red, _ = librosa.load(redhot_path)
```

```python
# Extract ZCR with librosa
FRAME_LENGTH = 1024
HOP_LENGTH = 512

zcr_deb = librosa.feature.zero_crossing_rate(y=deb, frame_length=FRAME_LENGTH, hop_length=HOP_LENGTH)[0]
zcr_duk = librosa.feature.zero_crossing_rate(y=duk, frame_length=FRAME_LENGTH, hop_length=HOP_LENGTH)[0]
zcr_red = librosa.feature.zero_crossing_rate(y=red, frame_length=FRAME_LENGTH, hop_length=HOP_LENGTH)[0]
```

```python
zcr_deb.shape
```

```python
frames = range(0, zcr_deb.size)
t = librosa.frames_to_time(frames, hop_length=HOP_LENGTH)

plt.figure(figsize=(15,17))
plt.plot(t, zcr_deb, color='y')
plt.plot(t, zcr_duk, color='b')
plt.plot(t, zcr_red, color='r')
plt.ylim(0, 0.5)

plt.show()
```
